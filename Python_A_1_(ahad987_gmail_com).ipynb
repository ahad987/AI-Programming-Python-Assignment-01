# **Question : 1**
Given a list of integers, extract all numbers that are divisible by 4 but not divisible by 8.
# Answer to the question No.: 1

```python
numbers = input("Enter numbers: ").split()

result = []

for num in numbers:
    num = int(num)

    if num % 4 == 0 and num % 8 != 0:
        result.append(num)

print(result)
```

```python
# Using list comprehension
numbers = [int(x) for x in input("Enter space-separated integers: ").split()]

print("The resulting list:", numbers)
```

# **Question : 2**
Write a function compute_ratio(x, y) that returns (x + y) / (x * y). Use exception handling to manage division by zero.
# Answer to the question No.: 2

```python
def compute_ratio(x, y):
  try:
    result = (x + y) / (x * y)
    return result
    # print("The result is:", result)
  except ZeroDivisionError:
    # print("Error: Division not possible.")
    return "Error: Division not possible"

x = float(input("Enter the value of x: "))
y = float(input("Enter the value of y: "))

print(compute_ratio(x, y))
```

# **Question : 3**
Write a Python program that reads a text file and finds the longest word.
# Answer to the question No.: 3

```python
with open("data.txt", "r") as file:
    text = file.read()

words = text.split()

longest_word = words[0]

for word in words:
  print(f"{word} = {len(word)}")
  if len(word) > len(longest_word):
    longest_word = word

print(longest_word)
```

# **Question : 4**
Write a program that converts a sentence into a list of words but only includes words that start with a vowel (a, e, i, o, u), ignoring case.
# Answer to the question No.: 4

```python
sentence = input("Enter a sentence: ")

sentence = sentence.lower()

words = sentence.split()

result = []

for word in words:
  if word[0] in "aeiou":
    result.append(word)

print(result)
print(words)
```

# **Question : 5**
A word is called a Palindrome if reversing its letters results in the same word (ignoring case). Write a program to check whether a given string is a palindrome.
# Answer to the question No.: 5

```python
word = input("Enter a word: ")
word = word.lower()
word2 = word[::-1]

if word == word2:
  print("Palindrome")
else:
  print("Not Palindrome")
```
